```python
import torch
a=torch.tensor([2,4,6,8])
print(a)
print(a.numel())
print(a.size())
print(a.ndim)
print(a.shape)
print(a.dtype)
print(torch.ones(2,3,dtype=int))
print(torch.zeros(2,2,dtype=int))
print(torch.arange(5,16,2))
print(torch.range(5,16,2))
```

```python
import torch
a=torch.tensor([[18,36,54,78],[36,94,72,81],[24,56,79,3],[1,2,3,4]])
print(a)
print(a[1])
print(a[2])
print(a[[2,3]])
print(a[[2,3],2])
print(a[[0,1,2],[1,2,3]])
print(a[2,1])
print(a[2:4,2:4])
print(a[1:3,1:3])
```

# operations

```python
import torch
a=torch.tensor([[[18,36,54,78],[36,94,72,81]],[[24,56,79,3],[1,2,3,4]]])
print(a)
```

```python
import torch
a=torch.tensor([[1,2],[5,6]])
b=torch.tensor([[7,8],[11,12]])
print(a,b)
print(torch.add(a,b))
print(torch.subtract(a,b))
print(torch.multiply(a,b))
print(torch.divide(a,b))
print(a@b)
print(torch.mul(a,b))
```

```python
import torch
a=torch.tensor([[1,2],[5,6]])
b=torch.tensor([[7,8],[11,12]])
print(a,b)
print(torch.flip(a,dims=[0]))
print(torch.flip(b,dims=[1]))
print(torch.flip(b,dims=[0,1]))
print(torch.flip(b,dims=[1,0]))
```

```python
import torch
a=torch.tensor([1,2,111,11,89])
print(a)
print(a.argmax())
print(a.max())
print(torch.max(a))
print(torch.argmax(a))
```

```python
import torch
a=torch.tensor([1,2,111,11,89])
print(a)
print(a.argmin())
print(a.min())
print(torch.min(a))
print(torch.argmin(a))
```

```python
import torch
AI=torch.tensor([40,60,30,70,95])
print(AI>=70)
print(AI[AI>=70])
print(AI[AI<35])
```
